## 空氣品質指標(AQI)概況
    - https://data.gov.tw/dataset/89029

```python
import requests

# 空氣品質指標 API（2023 年環保署改制為環境部，網址改為 data.moenv.gov.tw）
# 請先到「環境部環境資料開放平臺」https://data.moenv.gov.tw 註冊會員，在會員專區取得免費 API Key，貼在下方引號內
API_KEY = ''
CSV_URL = f'https://data.moenv.gov.tw/api/v2/aqx_p_432?api_key={API_KEY}&limit=1000&format=CSV'
# 還沒有 API Key 時，改用教材保存的一份環境部真實觀測資料（2021/11/08 18:00）
SAMPLE_URL = 'https://raw.githubusercontent.com/roberthsu2003/PythonForDataAnalysis/master/pandas%E7%9A%84%E6%AA%94%E6%A1%88%E5%AD%98%E5%8F%96/%E7%A9%BA%E6%B0%A3%E5%93%81%E8%B3%AA%E6%8C%87%E6%A8%99.csv'

#下載並存檔
response = requests.get(CSV_URL if API_KEY else SAMPLE_URL)
if response.status_code == 200 and 'API KEY' not in response.text[:100]:
    with open('空氣品質指標.csv','wb') as f:
        for chunk in response.iter_content(chunk_size=128):
            f.write(chunk)
    print('已下載：', '環境部 API 即時資料' if API_KEY else '教材保存的範例資料')
else:
    print('下載錯誤：請確認 API Key 是否正確')
```

```python
# 新版 API 的欄位名稱是小寫（例如 sitename），統一成舊版寫法，後面的程式才能沿用
standard_columns = ['SiteName', 'County', 'AQI', 'Pollutant', 'Status', 'PM2.5', 'PM2.5_AVG', 'PM10', 'PM10_AVG',
                    'PublishTime', 'SiteId', 'Longitude', 'Latitude', 'WindSpeed', 'WindDirec']
def to_standard(name):
    key = name.lower().replace('_', '')
    return next((s for s in standard_columns if s.lower().replace('_', '') == key), name)
import pandas as pd 
aqiDataFrame = pd.read_csv('空氣品質指標.csv').rename(columns=to_standard)
aqiDataFrame.dtypes
```

```python
aqi=aqiDataFrame.reindex(columns=['SiteName','AQI','PM2.5_AVG','PublishTime'])
aqi
```

```python
min = aqi['AQI'].min()
max = aqi['AQI'].max()
mean = aqi['AQI'].mean()
```

```python
#最好
aqi[aqi['AQI'] == min]
```

```python
#最不好
aqi[aqi['AQI'] == max]
```
