## 🧭 資料專案介紹：貫徹「看懂資料與選對圖表」思維 SOP

在動手向 AI 提問前，我們先依據先前所學的思維框架梳理這份資料：

### 1. 欄位角色貼標籤
- ⚪ **標識 (ID)**：`座號`、`姓名` ➔ 代表「這是誰」，**只能計數，絕對不能拿來計算加總或平均**！
- 🟢 **指標 (Metric)**：`國文`、`英文`、`數學`、`自然`、`社會` ➔ 連續數值（0~100 分），代表**衡量學生的學業表現**。

### 2. 五大提問積木與核心問題
- 🎯 **比較**：哪一科全班平均最高？哪一科最難？誰的總分排在前三名？誰需要補救？
- 📊 **分布**：最難科目的分數主要集中在哪個分數區間？有沒有極端低分？
- 🔗 **關係**：數學成績與自然成績是否存在連動正相關？

### 3. 邊界反向檢核防呆（不可觸犯的紅線）
- ⚠️ **缺考（空白/NaN）絕不能當成 0 分！** 若把缺考填 0 分，會嚴重拉低全班平均並扭曲排名。算平均時必須忽略缺考。
- ⚠️ **標識欄位不運算**：座號與姓名不可參與任何統計運算。
- ⚠️ **相關不推因果**：就算數學與自然高度相關，也不能推論為「數學好造成自然好」。

### 4. 決定最適圖表
- 各科平均比高低 ➔ **垂直長條圖**（少數類別，Y 軸數值從 0 開始）。
- 單科分數集中趨勢與極端值 ➔ **直方圖** 或 **箱形圖**。
- 雙科成績連動關聯 ➔ **散佈圖**（成對數值，點切勿連線）。

````python
# 第 0 步：下載範例資料並安裝繁體中文字型（請先點擊左側 ▶ 執行此儲存格）
import os
from pathlib import Path
from urllib.request import urlretrieve

# 1. 下載資料
data_dir = Path('data')
data_dir.mkdir(exist_ok=True)
csv_url = 'https://raw.githubusercontent.com/roberthsu2003/PythonForDataAnalysis/master/AI%E5%8D%94%E4%BD%9C%E8%B3%87%E6%96%99%E5%88%86%E6%9E%90_12%E5%80%8B%E7%AF%84%E4%BE%8B/01_class_scores/data/scores.csv'
urlretrieve(csv_url, data_dir / 'scores.csv')

# 2. 安裝與設定 Colab 繁體中文字型 (思源黑體)
%pip install -q mplfonts
import glob
import mplfonts
from matplotlib import font_manager
from mplfonts import use_font

for f in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(f)
use_font("Noto Sans CJK SC")

print("✅ 環境初始化完成！資料已就緒，繁體中文字型已設定成功。請開始進行下方單元。")
````

### 📝 單元一：讀取資料與欄位角色檢核（貼標籤）

> 🎯 **思維重點**：檢驗資料欄位，確認哪些是「標識 (ID)」、哪些是「指標 (Metric)」，並找出缺考空值。

📋 **複製以下 Prompt 貼給 Colab Gemini**：
```text
請幫我用 pandas 讀取 data/scores.csv：
1. 顯示前 5 筆資料與所有欄位名稱。
2. 檢查各欄位的資料型態。
3. 統計每個科目分別有多少人「缺考」（缺失值 NaN）。
4. 用文字回報：座號與姓名是否為識別標籤，五科成績是否為數值指標。
```

````python
# 👉 請將 Gemini 產生的程式碼貼於此處執行（或使用 Colab 內建 AI 直接產生於此）
import pandas as pd

df = pd.read_csv('data/scores.csv')
print("--- 前 5 筆資料 ---")
display(df.head())

print("\n--- 欄位型態與缺值統計 ---")
print(df.info())
print("\n各科目缺考人數：")
print(df.isna().sum())
````

### 📝 單元二：資料防呆與描述統計（邊界檢核）

> 🎯 **思維重點**：邊界反向檢核——缺考（空值）絕對不能當成 0 分！計算全班平均時必須忽略空值。

📋 **複製以下 Prompt 貼給 Colab Gemini**：
```text
請計算全班 5 科（國文、英文、數學、自然、社會）的統計數據：
1. 計算每科的平均分、最高分、最低分與標準差，四捨五入保留一位小數。
2. 計算平均時必須「忽略缺考（NaN）」，不可把缺考補 0 分，並在註解說明這樣做的原因。
3. 將 5 科的統計數據整理成美觀的表格顯示。
```

````python
# 👉 請將 Gemini 產生的程式碼貼於此處執行
subjects = ['國文', '英文', '數學', '自然', '社會']
stats_df = pd.DataFrame({
    '平均分': df[subjects].mean().round(1),
    '最高分': df[subjects].max(),
    '最低分': df[subjects].min(),
    '標準差': df[subjects].std().round(1)
})
display(stats_df)
````

### 📝 單元三：提問積木【比較】— 哪一科最難？（垂直長條圖）

> 🎯 **思維重點**：問題類別屬於「比較」，少數類別比高低，依據選圖規則挑選【垂直長條圖】，Y 軸數值必須從 0 開始！

📋 **複製以下 Prompt 貼給 Colab Gemini**：
```text
請用 matplotlib 或 seaborn 畫出 5 科平均成績的「垂直長條圖」：
1. X 軸為科目名稱，Y 軸為平均分數。
2. Y 軸數值起點必須從 0 開始，上限設為 100。
3. 在柱子上方標註具體平均分數數值（保留一位小數）。
4. 最低分的科目柱子用醒目的顏色（如深紅色或橘色）標示，並在標題標明「最難科目」。
5. 圖表標題、座標軸與標籤全部使用繁體中文。
```

````python
# 👉 請將 Gemini 產生的程式碼貼於此處執行
import matplotlib.pyplot as plt

means = stats_df['平均分']
min_subject = means.idxmin()
colors = ['#e74c3c' if sub == min_subject else '#3498db' for sub in subjects]

plt.figure(figsize=(8, 5))
bars = plt.bar(subjects, means, color=colors, width=0.55)

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 1.5, f'{yval:.1f}', ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.ylim(0, 105)
plt.title(f'全班五科平均分數比較（最難科目：{min_subject}）', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('科目', fontsize=12)
plt.ylabel('平均分數', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
````

### 📝 單元四：提問積木【分布】— 最難科目分數集中在哪裡？（直方圖與箱形圖）

> 🎯 **思維重點**：問題類別屬於「分布」，連續數值看集中區間用【直方圖】，秒抓極端離群值用【箱形圖】。

📋 **複製以下 Prompt 貼給 Colab Gemini**：
```text
針對全班平均最低的「數學」科，請畫出圖表來觀察分數分布：
1. 建立兩張並排的子圖：左邊畫「直方圖（Histogram）」觀察分數集中區間；右邊畫「箱形圖（Box Plot）」觀察中位數與是否有極端低分的離群值。
2. 直方圖請將分數區間（bins）設為適當大小，並畫出平均線。
3. 箱形圖標出五數概括與異常值點。
4. 全圖使用繁體中文標籤，並以文字說明全班數學表現是常態還是兩極分化。
```

````python
# 👉 請將 Gemini 產生的程式碼貼於此處執行
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 左圖：直方圖
axes[0].hist(df['數學'].dropna(), bins=8, color='#3498db', edgecolor='white', alpha=0.85)
axes[0].axvline(df['數學'].mean(), color='#e74c3c', linestyle='--', linewidth=2, label=f'平均分: {df["數學"].mean():.1f}')
axes[0].set_title('數學科分數分布直方圖', fontsize=13, fontweight='bold')
axes[0].set_xlabel('分數區間', fontsize=11)
axes[0].set_ylabel('人數', fontsize=11)
axes[0].legend()
axes[0].grid(axis='y', linestyle='--', alpha=0.4)

# 右圖：箱形圖
axes[1].boxplot(df['數學'].dropna(), vert=True, patch_artist=True, boxprops=dict(facecolor='#a8dadc', color='#1d3557'))
axes[1].set_title('數學科箱形圖（檢視離群值）', fontsize=13, fontweight='bold')
axes[1].set_ylabel('分數', fontsize=11)
axes[1].set_xticklabels(['數學科'])
axes[1].grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
````

### 📝 單元五：提問積木【關係】— 數學與自然成績是否有連動相關？（散佈圖）

> 🎯 **思維重點**：問題類別屬於「關係」，看兩個成對數值是否連動用【散佈圖】（點切勿連線！）。邊界防呆提醒：相關不推因果！

📋 **複製以下 Prompt 貼給 Colab Gemini**：
```text
請分析「數學」與「自然」這兩科的成績是否有連動關聯：
1. 繪製兩科的「散佈圖（Scatter Plot）」，X 軸為數學成績、Y 軸為自然成績。
2. 每個點代表一位學生，點與點之間「不要連線」。
3. 計算兩科的皮爾森相關係數（Pearson correlation），並將數值顯示於圖表標題。
4. 用文字解讀兩科是否存在正相關，並附註「相關性不等於因果關係」的防呆結論。
```

````python
# 👉 請將 Gemini 產生的程式碼貼於此處執行
import numpy as np

valid_data = df[['數學', '自然']].dropna()
corr = valid_data['數學'].corr(valid_data['自然'])

plt.figure(figsize=(7, 5))
plt.scatter(valid_data['數學'], valid_data['自然'], color='#2ecc71', edgecolors='#27ae60', s=60, alpha=0.8)

# 加上趨勢參考線
z = np.polyfit(valid_data['數學'], valid_data['自然'], 1)
p = np.poly1d(z)
plt.plot(valid_data['數學'], p(valid_data['數學']), color='#e74c3c', linestyle=':', label='趨勢參考線')

plt.title(f'數學 vs. 自然 成績相關性散佈圖（相關係數 r = {corr:.2f}）', fontsize=13, fontweight='bold')
plt.xlabel('數學成績', fontsize=11)
plt.ylabel('自然成績', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

print(f"💡 分析結論：數學與自然相關係數為 {corr:.2f}（顯著正相關）。\n⚠️ 邊界防呆：這僅代表擅長數學的同學通常也擅長自然，但絕不代表「補習數學就能自動讓自然進步」（相關 ≠ 因果）！")
````

### 📝 單元六：提問積木【比較】— 計算個人總分與排名，找出需補救學生

> 🎯 **思維重點**：計算學生總體表現與高低排名。邊界防呆注意：缺考學生總分應特別註記，不可直接列入不及格名單。

📋 **複製以下 Prompt 貼給 Colab Gemini**：
```text
請計算全班每位學生的成績表現：
1. 計算每位學生的總分（5 科加總）與平均分。若有缺考科目，總分需標明僅採計有考科目。
2. 依總分由高至低進行排名。
3. 列出全班總分前 3 名的優等生名單。
4. 找出平均低於 60 分或有任一科低於 50 分需要輔導補救的學生清單。
```

````python
# 👉 請將 Gemini 產生的程式碼貼於此處執行
df_result = df.copy()
df_result['總分'] = df_result[subjects].sum(axis=1)
df_result['平均'] = df_result[subjects].mean(axis=1).round(1)
df_result['排名'] = df_result['總分'].rank(ascending=False, method='min').astype(int)

df_sorted = df_result.sort_values('排名')

print("🏆 全班前 3 名：")
display(df_sorted[['排名', '座號', '姓名', '總分', '平均']].head(3))

needs_help = df_sorted[(df_sorted['平均'] < 60) | (df_sorted[subjects].min(axis=1) < 50)]
print(f"\n⚠️ 需重點輔導與補救教學學生（共 {len(needs_help)} 位）：")
display(needs_help[['排名', '座號', '姓名', '總分', '平均'] + subjects])
````
