## 🧭 資料專案介紹：貫徹「看懂資料與選對圖表」思維 SOP

本例不提供固定的資料檔。請從下表三份官方資料**只選一份**，在 Notebook 第 0 步把 `CHOICE` 改成對應代號，程式會從官方網址下載。下表是撰寫教材時（2026 年 10 月）實際下載核對的內容；**資料會更新，請以你下載當天的實際欄位為準**。

### 1. 欄位角色貼標籤
- 🔵 **條件 (Dimension)**：時間（A 的 `日期`；B、C 的 `民國年`＋`月份`）、類別（B 的 `毒品種類`／`觀勒次數`／`性別`；C 的 `被害補償金類別`）。
- 🟢 **指標 (Metric)**：B 的 `人`（人數）、C 的 `新臺幣元`（金額）、A 的 `實際收容`／`核定容額`（人數）➔ **人數、金額不能互換**。
- ⚪ **標識 (ID)**：這三份彙整統計沒有個人識別碼；也**不能**嘗試回推個人。

### 2. 五大提問積木與核心問題
- 🎯 **必做（依所選資料擇一）**：A【比較】實際收容與核定容額差多少？B【趨勢】每月新入所人數如何變化？C【比較】最近一個完整年度，各類別補償金額多少？
- 🧩 **組成（選做）**：A 男女組成；B 各毒品種類×觀勒次數的占比；C 各補償金類別占比——都只用**完整年度**比較。
- 🚫 **先寫下不能回答的問題**：例如「哪裡的人比較容易吸毒？」「政策實施後案件減少了多少是政策造成的？」

### 3. 邊界反向檢核防呆（不可觸犯的紅線）
- ⚠️ **先寫來源卡**：發布機關、資料頁網址、下載日期、資料期間、單位、一列代表什麼；來源不清楚就先停。
- ⚠️ **民國年要轉換、期間要完整**：民國 115 年＝西元 2026 年；最新年度若不滿 12 個月，不能和完整年度比較全年總量。
- ⚠️ **單日快照不能畫趨勢**：A 每次下載只有一天的資料，要看趨勢必須每天自行保存，不能用一筆資料說「越來越多」。
- ⚠️ **人數、件數、金額不能互換**：補償金額高不代表案件多；新入所人數不等於盛行率（沒有人口分母）。
- ⚠️ **彙整統計不推個人、不推政策因果**：不能回推任何個人，也不能把同期變化直接寫成某項政策的成效。
- ⚠️ **下載也可能失敗**：A 的官方網站憑證設定不完整，Python 自動下載會被拒絕；請改用瀏覽器下載後上傳，並在來源卡記下下載日期。
- ⚠️ **欄位會改版**：若實際欄位和教材不同，參考程式會停下來；請改用 Prompt 讓 AI 依實際欄位產生程式，**不得補造欄位或數字**。

### 4. 決定最適圖表
- A【比較】實際收容 vs 核定容額 ➔ **垂直長條圖**（Y 軸從 0 開始，標出資料日期與超收率）。
- B【趨勢】每月新入所人數 ➔ **折線圖**（X 軸由早到晚的年月）。
- C【比較】完整年度各類別金額 ➔ **垂直長條圖**（Y 軸從 0 開始，單位萬元）。
- 【組成，選做】➔ **百分比堆疊長條圖**（每個完整年度一根，寫明分母）；A 只有兩類可用單根 100% 長條。
- 每張圖都要在圖下方註明：資料來源、下載日期、資料期間與單位。

````python
# 第 0 步：選擇一份官方公開資料並下載（請先點擊左側 ▶ 執行此儲存格）
import os
from datetime import date
from pathlib import Path
from urllib.request import urlretrieve

# 1. 選一份來源：把 CHOICE 改成 'A'、'B' 或 'C'（一次只分析一份）
CHOICE = 'B'
SOURCES = {
    'A': {'名稱': '矯正機關每日收容動態', '頁面': 'https://data.gov.tw/dataset/101185',
          '下載': 'https://prisonmuseum.moj.gov.tw/jqw_pub/today.xml', '檔名': 'moj_daily_custody.xml'},
    'B': {'名稱': '成年勒戒處所新入所收容人人數按毒品種類及性別分', '頁面': 'https://data.gov.tw/dataset/39440',
          '下載': 'https://www.rjsd.moj.gov.tw/rjsdweb/OpenData.ashx?code=G08102', '檔名': 'moj_rehab_new_admissions.csv'},
    'C': {'名稱': '地方檢察署犯罪被害補償金決定補償金額（按補償金類別分）', '頁面': 'https://data.gov.tw/dataset/15128',
          '下載': 'https://www.rjsd.moj.gov.tw/rjsdweb/OpenData.ashx?code=QC1004', '檔名': 'moj_victim_compensation.csv'},
}
source = SOURCES[CHOICE]
data_dir = Path('data')
data_dir.mkdir(exist_ok=True)
data_path = data_dir / source['檔名']
download_date = '（手動上傳，請自行填寫下載日期）'
try:
    urlretrieve(source['下載'], data_path)
    download_date = date.today().isoformat()
    print(f"✅ 已下載：{source['名稱']} → {data_path}（下載日期 {download_date}）")
except Exception as error:
    print(f"⚠️ 自動下載失敗（{type(error).__name__}）。請到 {source['頁面']} 用瀏覽器下載，上傳到 data/ 並命名為 {source['檔名']}")
    print("   （例如 A 的官方網站憑證設定不完整，Python 會拒絕連線，但瀏覽器通常可以下載）")

# 2. 安裝與設定 Colab 繁體中文字型 (思源黑體)
%pip install -q mplfonts
import glob
import mplfonts
from matplotlib import font_manager
from mplfonts import use_font

for f in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(f)
use_font("Noto Sans CJK SC")

print("✅ 環境初始化完成！資料已就緒，繁體中文字型已設定成功。請開始進行下方單元。")
````

### 📝 單元一【必做】：來源卡與欄位角色（貼標籤）

> 🎯 **思維重點**：真實資料沒有教材資料卡，你要自己寫「來源卡」。讀檔後先看欄位、型態與缺值，再判斷一列代表什麼。

📋 **複製以下 Prompt 貼給 Colab Gemini**：
```text
我在 data/ 放了一份法務部公開統計（檔名見第 0 步），請幫我用 pandas 讀取：
1. CSV 請用 encoding="utf-8-sig"、把「月份」讀成文字；XML 請用 pd.read_xml。
2. 顯示前 5 筆資料、所有欄位名稱、列數、資料型態與各欄缺值數。
3. 幫我整理「來源卡」：資料名稱、發布機關、資料頁網址、下載日期、資料期間、單位、一列代表什麼。
4. 逐欄標記：標識、條件（時間／類別）、指標；特別說明哪個欄位是人數、哪個是金額。
如果欄位和我預期的不同，請照實回報，不要補造欄位。先不要畫圖。
```

````python
# 👉 請將 Gemini 產生的程式碼貼於此處執行
import pandas as pd

if not data_path.exists():
    raise FileNotFoundError(f"找不到 {data_path}：請先完成第 0 步；自動下載失敗時，請手動上傳並依提示命名。")
if data_path.suffix == '.xml':
    df = pd.read_xml(data_path, parser='etree')
else:
    df = pd.read_csv(data_path, encoding='utf-8-sig', dtype={'月份': str})

print(f"資料：{source['名稱']}\n資料頁：{source['頁面']}\n下載日期：{download_date}")
print(f"列數 {len(df)}、欄數 {df.shape[1]}")
display(df.head())
df.info()
print("各欄缺值：", df.isna().sum().to_dict())

# 撰寫教材時（2026 年 10 月）核對的欄位；若官方改版，以下單元會停下來提醒
EXPECTED = {
    'A': ['日期', '實際收容', '男', '女', '核定容額', '超收率'],
    'B': ['毒品種類', '觀勒次數', '性別', '民國年', '月份', '人'],
    'C': ['被害補償金類別', '民國年', '月份', '新臺幣元'],
}
missing = [c for c in EXPECTED[CHOICE] if c not in df.columns]
columns_ok = not missing
print("✅ 欄位與教材相同，可對照參考程式" if columns_ok else f"⚠️ 缺少欄位 {missing}：官方可能已改版，請改用 Prompt 讓 AI 依實際欄位產生程式")
````

### 📝 單元二【必做】：邊界檢核：民國年、資料期間、單日快照與總計列

> 🎯 **思維重點**：真實資料的陷阱要自己找：民國年要轉西元、最新年度是否不滿 12 個月、是否只有單日快照、是否混有「總計」列。

📋 **複製以下 Prompt 貼給 Colab Gemini**：
```text
請幫我檢查這份資料的分析邊界，並逐項回報：
1. 若有「民國年」與「月份」：新增西元的「年月」日期欄（民國年 + 1911），印出資料期間（最早～最晚），並列出每個民國年有幾個月份；不滿 12 個月的年度要特別標示。
2. 若只有「日期」欄：印出資料日期，說明是否只有單日快照、能不能畫時間趨勢。
3. 檢查是否有「總計」「合計」之類的列，以及完全重複的列。
4. 列出這份資料「可以回答」與「不能回答」的問題各一個（例如人數不等於盛行率、金額不等於件數）。
不要修改原始檔。
```

````python
# 👉 請將 Gemini 產生的程式碼貼於此處執行
warnings = []
if {'民國年', '月份'} <= set(df.columns):
    df['年月'] = pd.to_datetime((df['民國年'].astype(int) + 1911).astype(str) + '-' + df['月份'].astype(str).str.zfill(2) + '-01')
    months = df.groupby('民國年')['月份'].nunique()
    full_years = months[months == 12].index
    print(f"資料期間：{df['年月'].min():%Y-%m} ～ {df['年月'].max():%Y-%m}（西元）")
    print("各民國年的月份數：", months.to_dict())
    if (months < 12).any():
        warnings.append(f"民國 {list(months[months < 12].index)} 年不滿 12 個月，不能和完整年度比較全年總量")
if '日期' in df.columns:
    print("資料日期：", df['日期'].tolist())
    if len(df) == 1:
        warnings.append("只有單日快照，不能畫時間趨勢；資料日期可能早於下載日期")

has_total = df.astype(str).apply(lambda col: col.str.contains('總計|合計')).any(axis=1)
print(f"疑似總計列：{has_total.sum()} 列；完全重複列：{df.duplicated().sum()} 列")
for w in warnings:
    print("⚠️", w)
````

### 📝 單元三【必做】：一個問題、一張圖（依所選資料選圖）

> 🎯 **思維重點**：依所選資料只回答一個問題：A 比較（長條圖）、B 趨勢（折線圖）、C 比較完整年度（長條圖）。先產出數字表再畫圖，圖下方註明來源、下載日期與單位。

📋 **複製以下 Prompt 貼給 Colab Gemini**：
```text
請依我選的資料回答一個問題，先產出數字表，再畫圖：
- 若是 A（每日收容動態）：比較「實際收容」與「核定容額」，畫垂直長條圖（Y 軸從 0 開始），標題寫出資料日期與超收率。
- 若是 B（勒戒新入所人數）：把所有毒品種類與性別加總成每月新入所人數，畫折線圖（X 軸為西元年月，由早到晚）。
- 若是 C（決定補償金額）：取最近一個「完整 12 個月」的年度，依補償金類別加總金額（換算成萬元），畫垂直長條圖（Y 軸從 0 開始）。
圖表文字使用繁體中文，圖下方註明：資料來源、下載日期、資料期間、單位。若欄位和描述不同，請依實際欄位調整並先告訴我。
```

````python
# 👉 請將 Gemini 產生的程式碼貼於此處執行
import matplotlib.pyplot as plt

def add_source_note():
    plt.figtext(0.01, -0.02, f"資料來源：法務部，{source['頁面']}；下載日期：{download_date}", fontsize=9, color='gray')

if not columns_ok:
    print("⚠️ 欄位與教材不同，參考程式停止。請改用上方 Prompt，讓 AI 依實際欄位產生程式。")
elif CHOICE == 'A':
    row = df.iloc[0]
    result = pd.DataFrame({'項目': ['核定容額', '實際收容'], '人數': [int(row['核定容額']), int(row['實際收容'])]})
    display(result)
    plt.figure(figsize=(6, 5))
    bars = plt.bar(result['項目'], result['人數'], color=['#95a5a6', '#e74c3c'], width=0.5)
    for bar in bars:
        plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height() * 1.01, f'{bar.get_height():,.0f}', ha='center')
    plt.ylim(0, result['人數'].max() * 1.15)
    plt.title(f"矯正機關實際收容與核定容額（{row['日期']}，超收率 {row['超收率']}）", fontsize=12, fontweight='bold')
    plt.ylabel('人數（人）')
    add_source_note()
    plt.show()
    print("⚠️ 這是單日快照，不能據此說收容人數「越來越多」。")
elif CHOICE == 'B':
    result = df.groupby('年月', as_index=False)['人'].sum().rename(columns={'人': '新入所人數'})
    display(result.tail(12))
    plt.figure(figsize=(11, 5))
    plt.plot(result['年月'], result['新入所人數'], marker='o', markersize=3, color='#3498db', linewidth=2)
    plt.ylim(0, result['新入所人數'].max() * 1.15)
    plt.title(f"成年勒戒處所每月新入所人數（{result['年月'].min():%Y-%m}～{result['年月'].max():%Y-%m}，所有毒品種類與性別合計）",
              fontsize=12, fontweight='bold')
    plt.xlabel('年月（西元）')
    plt.ylabel('新入所人數（人）')
    plt.grid(linestyle='--', alpha=0.4)
    add_source_note()
    plt.show()
    print("⚠️ 新入所人數不等於盛行率；人數變化也不能直接說成某項政策的成效。")
elif CHOICE == 'C':
    latest = max(full_years)
    result = (df[df['民國年'] == latest].groupby('被害補償金類別', as_index=False)['新臺幣元'].sum()
              .sort_values('新臺幣元', ascending=False))
    result['萬元'] = (result['新臺幣元'] / 10000).round(1)
    display(result)
    plt.figure(figsize=(8, 5))
    bars = plt.bar(result['被害補償金類別'], result['萬元'], color='#3498db', width=0.5)
    for bar in bars:
        plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height() * 1.01, f'{bar.get_height():,.0f}', ha='center')
    plt.ylim(0, result['萬元'].max() * 1.15)
    plt.title(f"民國 {latest} 年（西元 {latest + 1911} 年）各類別決定補償金額", fontsize=12, fontweight='bold')
    plt.ylabel('決定補償金額（萬元）')
    add_source_note()
    plt.show()
    print("⚠️ 金額不是件數或人數，不能據此排名哪類案件比較多或比較嚴重。")
````

### 📝 單元五【選做挑戰】：提問積木【組成】— 同一份資料再問一題（百分比堆疊長條圖）

> 🎯 **思維重點**：換一種提問積木，但不換資料。組成問題先寫分母，而且只用**完整年度**比較；不跨資料集硬拼不同口徑。

📋 **複製以下 Prompt 貼給 Colab Gemini**：
```text
請在同一份資料上做一題「組成」分析：
- 若是 A：實際收容中男、女各占多少？畫一根 100% 水平長條，分母為當日實際收容人數。
- 若是 B：每個完整年度中，「毒品種類 × 觀勒次數」四種組合各占新入所人數多少？
- 若是 C：每個完整年度中，各補償金類別占決定補償金額多少？
B、C 只使用滿 12 個月的年度；畫百分比堆疊長條圖（每個年度一根，總和 100%），標題寫明分母。最後用一句話描述組成的變化，並說明不能推論的事。
```

````python
# 👉 請將 Gemini 產生的程式碼貼於此處執行
if not columns_ok:
    print("⚠️ 欄位與教材不同，請改用 Prompt 讓 AI 依實際欄位產生程式。")
elif CHOICE == 'A':
    row = df.iloc[0]
    share = pd.Series({'男': int(row['男']), '女': int(row['女'])})
    pct = share / share.sum() * 100
    fig, ax = plt.subplots(figsize=(9, 2.2))
    ax.barh(0, pct['男'], color='#5dade2')
    ax.barh(0, pct['女'], left=pct['男'], color='#f1948a')
    ax.text(pct['男'] / 2, 0, f"男 {pct['男']:.1f}%", ha='center', va='center')
    ax.text(pct['男'] + pct['女'] / 2, 0, f"女 {pct['女']:.1f}%", ha='center', va='center')
    ax.set_xlim(0, 100)
    ax.set_yticks([])
    ax.set_title(f"實際收容男女組成（{row['日期']}，分母＝實際收容 {int(row['實際收容']):,} 人）", fontweight='bold')
    plt.show()
else:
    category = df['毒品種類'] + '・' + df['觀勒次數'] if CHOICE == 'B' else df['被害補償金類別']
    value_col = '人' if CHOICE == 'B' else '新臺幣元'
    full = df[df['民國年'].isin(full_years)].assign(類別=category)
    mix = full.pivot_table(index='民國年', columns='類別', values=value_col, aggfunc='sum')
    mix_pct = mix.div(mix.sum(axis=1), axis=0) * 100
    mix_pct.index = [f'{y}年（{y + 1911}）' for y in mix_pct.index]
    display(mix_pct.round(1))
    ax = mix_pct.plot.barh(stacked=True, figsize=(10, 4.5), colormap='Set2', width=0.7)
    ax.set_xlim(0, 100)
    ax.set_xlabel('占該年度合計（%）')
    ax.set_ylabel('民國年（西元）')
    ax.set_title(f"各完整年度的組成（分母＝該年度{'新入所人數' if CHOICE == 'B' else '決定補償金額'}合計）", fontweight='bold')
    ax.legend(title='類別', bbox_to_anchor=(1.01, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
print("⚠️ 組成的變化只描述占比，不代表某一類「增加」或「減少」的原因。")
````
