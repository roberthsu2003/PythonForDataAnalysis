本例使用少量**課堂示意資料**，直接寫在程式中。依序執行兩格程式碼：先設定繁體中文字型，再顯示資料表並畫圖。

```python
%pip install -q mplfonts

import glob
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import mplfonts
from matplotlib import font_manager
from mplfonts import use_font

# 沿用本專案其他 Colab 範例的中文字型設定；此字型可顯示繁體中文。
for font_file in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(font_file)
sns.set_theme(style="whitegrid")
use_font("Noto Sans CJK SC")
plt.rcParams["axes.unicode_minus"] = False
print("✅ 繁體中文圖表環境準備完成")
```

```python
values = {"甲組": [5, 6, 6, 7, 7, 8],
          "乙組": [7, 8, 8, 9, 9, 20],
          "丙組": [4, 5, 5, 6, 6, 7]}
df = pd.DataFrame([(team, hour) for team, hours in values.items() for hour in hours],
                  columns=["團隊", "處理時間（小時）"])
display(df)
fig, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(data=df, x="團隊", y="處理時間（小時）", order=["甲組", "乙組", "丙組"],
            color="#60a5fa", ax=ax)
ax.set(title="乙組有一筆明顯較長的客服處理時間", ylim=(0, 23))
plt.tight_layout()
plt.show()
```

## 看圖表說商業分析

**觀察：**丙組中位數 5.5 小時、甲組 6.5 小時、乙組 8.5 小時；乙組另有 20 小時的高值。

**商業解讀：**先查看乙組該案是否屬於特殊客訴，再決定是否調整流程或人力。

**限制：**每組只有 6 筆，且案件難度可能不同，不能直接當作團隊績效排名。
