**何時選這張圖：**項目較多時，以點與細線減少視覺負擔。

本例使用少量**課堂示意資料**，直接寫在程式中。依序執行兩格程式碼：先設定繁體中文字型，再顯示資料表並畫圖。

```python
%pip install -q mplfonts

import glob
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import mplfonts
from matplotlib import font_manager
from mplfonts import use_font

# 沿用本專案其他 Colab 範例的中文字型設定；此字型可顯示繁體中文。
for font_file in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(font_file)
sns.set_theme(style="whitegrid")
use_font("Noto Sans CJK SC")
plt.rcParams["axes.unicode_minus"] = False
print("✅ 繁體中文圖表環境準備完成")
```

```python
df = pd.DataFrame({"門市": [f"門市 {i:02d}" for i in range(1, 13)],
                   "銷售額（萬元）": [95, 90, 87, 80, 76, 72, 69, 62, 58, 53, 48, 41]})
df = df.sort_values("銷售額（萬元）")
display(df)
fig, ax = plt.subplots(figsize=(9, 6))
y = np.arange(len(df))
ax.hlines(y, 0, df["銷售額（萬元）"], color="#bfdbfe", linewidth=2)
ax.scatter(df["銷售額（萬元）"], y, s=85, color="#2563eb", zorder=3)
ax.set(yticks=y, yticklabels=df["門市"], xlabel="銷售額（萬元）", xlim=(0, 105),
       title="十二家門市的年度銷售額排名")
for yi, value in zip(y, df["銷售額（萬元）"]):
    ax.text(value + 2, yi, str(value), va="center")
plt.tight_layout()
plt.show()
```

## 看圖表說商業分析

**觀察：**門市 01 為 95 萬元，門市 12 為 41 萬元，差 54 萬元。

**商業解讀：**先比較兩店的客流、商圈與營業天數，再決定是否複製高績效門市的做法。

**限制：**單看銷售額，不能評估坪效或獲利。
