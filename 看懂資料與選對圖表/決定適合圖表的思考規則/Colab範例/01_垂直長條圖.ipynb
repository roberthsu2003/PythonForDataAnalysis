**何時選這張圖：**類別少、名稱短；數值軸從 0 開始。

本例使用少量**課堂示意資料**，直接寫在程式中。依序執行兩格程式碼：先設定繁體中文字型，再顯示資料表並畫圖。

```python
%pip install -q mplfonts

import glob
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import mplfonts
from matplotlib import font_manager
from mplfonts import use_font

# 沿用本專案其他 Colab 範例的中文字型設定；此字型可顯示繁體中文。
for font_file in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(font_file)
sns.set_theme(style="whitegrid")
use_font("Noto Sans CJK SC")
plt.rcParams["axes.unicode_minus"] = False
print("✅ 繁體中文圖表環境準備完成")
```

```python
df = pd.DataFrame({"部門": ["行政", "財務", "業務", "資訊", "人資"],
                   "每人平均支出（萬元）": [120, 155, 185, 140, 165]})
display(df)
fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(df["部門"], df["每人平均支出（萬元）"],
              color=["#93c5fd", "#60a5fa", "#2563eb", "#60a5fa", "#93c5fd"])
ax.bar_label(bars, padding=3)
ax.set(title="業務部每人平均支出最高", xlabel="部門", ylabel="每人平均支出（萬元）", ylim=(0, 210))
plt.tight_layout()
plt.show()
```

## 看圖表說商業分析

**觀察：**業務部 185 萬元最高，行政部 120 萬元最低，差 65 萬元。

**商業解讀：**可先檢查業務部的支出結構，再判斷差異是否符合工作需求。

**限制：**支出高不等於浪費；還需看人員職務、產出與支出用途。
