**何時選這張圖：**X、Y 軸放兩個指標；圓面積代表第三個指標。

本例使用少量**課堂示意資料**，直接寫在程式中。依序執行兩格程式碼：先設定繁體中文字型，再顯示資料表並畫圖。

```python
%pip install -q mplfonts

import glob
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import mplfonts
from matplotlib import font_manager
from mplfonts import use_font

# 沿用本專案其他 Colab 範例的中文字型設定；此字型可顯示繁體中文。
for font_file in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(font_file)
sns.set_theme(style="whitegrid")
use_font("Noto Sans CJK SC")
plt.rcParams["axes.unicode_minus"] = False
print("✅ 繁體中文圖表環境準備完成")
```

```python
df = pd.DataFrame({"門市": list("ABCDEF"),
                   "廣告費（萬元）": [20, 30, 40, 50, 60, 70],
                   "營收（萬元）": [60, 75, 90, 100, 125, 135],
                   "利潤（萬元）": [8, 12, 7, 18, 15, 22]})
display(df)
fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(df["廣告費（萬元）"], df["營收（萬元）"],
           s=df["利潤（萬元）"] * 35, alpha=0.55, color="#2563eb", edgecolor="#1e3a8a")
for row in df.itertuples(index=False):
    ax.annotate(row.門市, (row[1], row[2]), ha="center", va="center", weight="bold")
ax.set(title="門市 F 利潤最高，氣泡面積代表利潤", xlabel="廣告費（萬元）", ylabel="營收（萬元）")
plt.tight_layout()
plt.show()
```

## 看圖表說商業分析

**觀察：**門市 F 的利潤 22 萬元最高；門市 C 雖有 90 萬元營收，利潤只有 7 萬元。

**商業解讀：**比較門市時要同時看營收與利潤，避免只追求高營收。

**限制：**氣泡面積較難精確比大小；需要精準決策時，仍應讀取原始數值表。
