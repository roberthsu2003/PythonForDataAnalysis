**何時選這張圖：**每季都以本季總營收為 100%；跨季比較結構。

本例使用少量**課堂示意資料**，直接寫在程式中。依序執行兩格程式碼：先設定繁體中文字型，再顯示資料表並畫圖。

```python
%pip install -q mplfonts

import glob
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import mplfonts
from matplotlib import font_manager
from mplfonts import use_font

# 沿用本專案其他 Colab 範例的中文字型設定；此字型可顯示繁體中文。
for font_file in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(font_file)
sns.set_theme(style="whitegrid")
use_font("Noto Sans CJK SC")
plt.rcParams["axes.unicode_minus"] = False
print("✅ 繁體中文圖表環境準備完成")
```

```python
df = pd.DataFrame({"季度": ["第 1 季", "第 2 季", "第 3 季", "第 4 季"],
                   "網站": [35, 45, 50, 55], "實體店": [30, 25, 25, 20],
                   "外送": [20, 20, 15, 15], "企業客戶": [15, 10, 10, 10]})
display(df)
fig, ax = plt.subplots(figsize=(9, 5))
left = np.zeros(len(df))
for column, color in zip(["網站", "實體店", "外送", "企業客戶"],
                         ["#1d4ed8", "#3b82f6", "#93c5fd", "#cbd5e1"]):
    values = df[column].to_numpy()
    ax.barh(df["季度"], values, left=left, label=column, color=color)
    for i, value in enumerate(values):
        text_color = "white" if column in ["網站", "實體店"] else "#0f172a"
        ax.text(left[i] + value / 2, i, f"{value}%", ha="center", va="center", color=text_color)
    left += values
ax.invert_yaxis()
ax.set(title="網站通路占比逐季提高", xlabel="各季營收占比（%）", xlim=(0, 100))
ax.legend(title="通路", loc="center left", bbox_to_anchor=(1.01, 0.5))
plt.tight_layout()
plt.show()
```

## 看圖表說商業分析

**觀察：**網站通路從第 1 季的 35% 增至第 4 季的 55%，增加 20 個百分點。

**商業解讀：**可檢查線上客服、庫存與物流容量是否跟上通路結構變化。

**限制：**百分比上升不一定代表網站營收絕對金額增加；每季分母不同，需再看各季總額。
