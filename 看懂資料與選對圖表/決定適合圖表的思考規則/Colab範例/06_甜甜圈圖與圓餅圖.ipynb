本例使用少量**課堂示意資料**，直接寫在程式中。依序執行兩格程式碼：先設定繁體中文字型，再顯示資料表並畫圖。

```python
%pip install -q mplfonts

import glob
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import mplfonts
from matplotlib import font_manager
from mplfonts import use_font

# 沿用本專案其他 Colab 範例的中文字型設定；此字型可顯示繁體中文。
for font_file in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(font_file)
sns.set_theme(style="whitegrid")
use_font("Noto Sans CJK SC")
plt.rcParams["axes.unicode_minus"] = False
print("✅ 繁體中文圖表環境準備完成")
```

```python
df = pd.DataFrame({"收入來源": ["基本方案", "進階方案", "企業方案", "加購服務"],
                   "收入（萬元）": [50, 25, 15, 10]})
display(df)
fig, ax = plt.subplots(figsize=(7, 6))
ax.pie(df["收入（萬元）"], labels=df["收入來源"], autopct="%1.0f%%",
       startangle=90, colors=["#1d4ed8", "#3b82f6", "#93c5fd", "#cbd5e1"],
       wedgeprops={"width": 0.42, "edgecolor": "white"}, textprops={"fontsize": 12})
ax.text(0, 0, "總收入\n100 萬元", ha="center", va="center", fontsize=13, weight="bold")
ax.set_title("基本方案占訂閱收入一半")
plt.tight_layout()
plt.show()
```

## 看圖表說商業分析

**觀察：**基本方案收入 50 萬元，占總收入 100 萬元的 50%。

**商業解讀：**基本方案是目前主力，應關注續訂率，也可研究是否有升級進階方案的空間。

**限制：**占比高不代表利潤率最高，仍需搭配各方案服務成本。
