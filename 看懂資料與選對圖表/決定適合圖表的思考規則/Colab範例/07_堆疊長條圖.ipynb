本例使用少量**課堂示意資料**，直接寫在程式中。依序執行兩格程式碼：先設定繁體中文字型，再顯示資料表並畫圖。

```python
%pip install -q mplfonts

import glob
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import mplfonts
from matplotlib import font_manager
from mplfonts import use_font

# 沿用本專案其他 Colab 範例的中文字型設定；此字型可顯示繁體中文。
for font_file in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(font_file)
sns.set_theme(style="whitegrid")
use_font("Noto Sans CJK SC")
plt.rcParams["axes.unicode_minus"] = False
print("✅ 繁體中文圖表環境準備完成")
```

```python
df = pd.DataFrame({"區域": ["北區", "中區", "南區"],
                   "飲品": [40, 32, 28], "點心": [25, 24, 20], "周邊": [10, 8, 7]})
display(df)
fig, ax = plt.subplots(figsize=(8, 5))
bottom = np.zeros(len(df))
for column, color in zip(["飲品", "點心", "周邊"], ["#1d4ed8", "#60a5fa", "#cbd5e1"]):
    ax.bar(df["區域"], df[column], bottom=bottom, label=column, color=color)
    bottom += df[column].to_numpy()
for i, total in enumerate(bottom):
    ax.text(i, total + 1, f"{total:.0f}", ha="center")
ax.set(title="北區總銷售額最高，飲品是主要來源", ylabel="銷售額（萬元）", ylim=(0, 85))
ax.legend(title="品類")
plt.tight_layout()
plt.show()
```

## 看圖表說商業分析

**觀察：**北區總額 75 萬元，中區 64 萬元，南區 55 萬元；三區都以飲品為最大來源。

**商業解讀：**南區可以先查飲品的客流與供應狀況。

**限制：**堆疊圖上方色塊基準線不同，若要精確比較各區點心銷售額，另畫分組長條圖更清楚。
