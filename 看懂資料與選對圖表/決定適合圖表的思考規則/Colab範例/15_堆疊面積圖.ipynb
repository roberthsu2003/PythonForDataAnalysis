**何時選這張圖：**類別少，兼看總量走勢與各類別貢獻。

本例使用少量**課堂示意資料**，直接寫在程式中。依序執行兩格程式碼：先設定繁體中文字型，再顯示資料表並畫圖。

```python
%pip install -q mplfonts

import glob
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import mplfonts
from matplotlib import font_manager
from mplfonts import use_font

# 沿用本專案其他 Colab 範例的中文字型設定；此字型可顯示繁體中文。
for font_file in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(font_file)
sns.set_theme(style="whitegrid")
use_font("Noto Sans CJK SC")
plt.rcParams["axes.unicode_minus"] = False
print("✅ 繁體中文圖表環境準備完成")
```

```python
df = pd.DataFrame({"月份": [1, 2, 3, 4, 5, 6],
                   "飲品": [30, 34, 33, 39, 42, 45],
                   "點心": [18, 19, 20, 22, 24, 26],
                   "周邊": [8, 9, 9, 10, 11, 12]})
display(df)
fig, ax = plt.subplots(figsize=(9, 5))
ax.stackplot(df["月份"], df["飲品"], df["點心"], df["周邊"],
             labels=["飲品", "點心", "周邊"], colors=["#1d4ed8", "#60a5fa", "#bfdbfe"])
ax.set(title="三種商品合計營收逐月增加", xlabel="月份", ylabel="營收（萬元）",
       xticks=df["月份"], xlim=(1, 6), ylim=(0, 95))
ax.legend(title="商品類別", loc="upper left")
plt.tight_layout()
plt.show()
```

## 看圖表說商業分析

**觀察：**三類商品合計從 1 月的 56 萬元增加到 6 月的 83 萬元，多 27 萬元；飲品一直是最大來源。

**商業解讀：**整體備貨量需要跟隨總營收成長，飲品供應尤其重要。
