**何時選這張圖：**把連續時間切成區間，觀察集中與尾端。

本例使用少量**課堂示意資料**，直接寫在程式中。依序執行兩格程式碼：先設定繁體中文字型，再顯示資料表並畫圖。

```python
%pip install -q mplfonts

import glob
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import mplfonts
from matplotlib import font_manager
from mplfonts import use_font

# 沿用本專案其他 Colab 範例的中文字型設定；此字型可顯示繁體中文。
for font_file in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(font_file)
sns.set_theme(style="whitegrid")
use_font("Noto Sans CJK SC")
plt.rcParams["axes.unicode_minus"] = False
print("✅ 繁體中文圖表環境準備完成")
```

```python
times = [15, 18, 20, 22, 25, 27, 28, 30, 31, 32, 34, 35,
         36, 37, 38, 40, 41, 42, 45, 47, 50, 55, 70]
df = pd.DataFrame({"配送分鐘": times})
display(df)
fig, ax = plt.subplots(figsize=(8, 5))
sns.histplot(data=df, x="配送分鐘", bins=np.arange(10, 81, 10), color="#2563eb", ax=ax)
ax.axvline(df["配送分鐘"].median(), color="#f97316", linestyle="--",
           label=f"中位數 {df['配送分鐘'].median():.0f} 分鐘")
ax.set(title="配送時間多集中在 20 至 50 分鐘", xlabel="配送時間（分鐘）", ylabel="訂單數")
ax.legend()
plt.tight_layout()
plt.show()
```

## 看圖表說商業分析

**觀察：**這 23 筆訂單的配送時間中位數為 35 分鐘，超過 50 分鐘的有 2 筆。

**商業解讀：**可檢查長時間配送的路線、時段或備貨流程。

**限制：**直方圖顯示區間分布；更換區間寬度會改變外觀，不能只憑圖形判定原因。
