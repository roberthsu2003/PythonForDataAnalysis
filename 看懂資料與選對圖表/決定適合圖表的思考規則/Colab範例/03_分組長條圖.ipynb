**何時選這張圖：**同一部門內並排比較兩個指標。

本例使用少量**課堂示意資料**，直接寫在程式中。依序執行兩格程式碼：先設定繁體中文字型，再顯示資料表並畫圖。

```python
%pip install -q mplfonts

import glob
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import mplfonts
from matplotlib import font_manager
from mplfonts import use_font

# 沿用本專案其他 Colab 範例的中文字型設定；此字型可顯示繁體中文。
for font_file in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(font_file)
sns.set_theme(style="whitegrid")
use_font("Noto Sans CJK SC")
plt.rcParams["axes.unicode_minus"] = False
print("✅ 繁體中文圖表環境準備完成")
```

```python
df = pd.DataFrame({"部門": ["行政", "財務", "資訊", "業務"],
                   "預算（萬元）": [100, 80, 120, 90],
                   "實際（萬元）": [108, 76, 135, 84]})
display(df)
x = np.arange(len(df))
fig, ax = plt.subplots(figsize=(9, 5))
bars1 = ax.bar(x - 0.18, df["預算（萬元）"], 0.36, label="預算", color="#93c5fd")
bars2 = ax.bar(x + 0.18, df["實際（萬元）"], 0.36, label="實際", color="#2563eb")
ax.bar_label(bars1, padding=2)
ax.bar_label(bars2, padding=2)
ax.set(xticks=x, xticklabels=df["部門"], ylabel="金額（萬元）", ylim=(0, 160),
       title="資訊部實際花費超過預算 15 萬元")
ax.legend(title="指標")
plt.tight_layout()
plt.show()
```

## 看圖表說商業分析

**觀察：**資訊部預算 120 萬元、實際 135 萬元，超支 15 萬元；財務部實際 76 萬元，比預算少 4 萬元。

**商業解讀：**資訊部可先拆解超支項目，確認是一次性設備更新還是持續性費用。

**限制：**「低於預算」也可能是計畫延後，不能直接視為績效更好。
