**何時選這張圖：**時間點多，按月份順序連線。

本例使用少量**課堂示意資料**，直接寫在程式中。依序執行兩格程式碼：先設定繁體中文字型，再顯示資料表並畫圖。

```python
%pip install -q mplfonts

import glob
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import mplfonts
from matplotlib import font_manager
from mplfonts import use_font

# 沿用本專案其他 Colab 範例的中文字型設定；此字型可顯示繁體中文。
for font_file in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(font_file)
sns.set_theme(style="whitegrid")
use_font("Noto Sans CJK SC")
plt.rcParams["axes.unicode_minus"] = False
print("✅ 繁體中文圖表環境準備完成")
```

```python
df = pd.DataFrame({"月份": np.arange(1, 13),
                   "營收（萬元）": [80, 85, 82, 90, 95, 100, 96, 105, 110, 108, 118, 130]})
display(df)
fig, ax = plt.subplots(figsize=(9, 5))
sns.lineplot(data=df, x="月份", y="營收（萬元）", marker="o", linewidth=2.5,
             color="#2563eb", ax=ax)
ax.set(title="營收整體上升，12 月達到高點", xticks=np.arange(1, 13), xlim=(1, 12))
plt.tight_layout()
plt.show()
```

## 看圖表說商業分析

**觀察：**1 月 80 萬元，12 月 130 萬元，增加 50 萬元、約 62.5%；中間仍有短暫回落。

**商業解讀：**可檢查年底高峰是否來自促銷、節慶或新增客戶，作為備貨參考。

**限制：**只有一年的資料，不能單憑這條線判斷每年都會重複相同季節型態。
