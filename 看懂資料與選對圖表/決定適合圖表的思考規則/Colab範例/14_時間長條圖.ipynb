**何時選這張圖：**時間點少；依時間順序排列，長條從 0 開始。

本例使用少量**課堂示意資料**，直接寫在程式中。依序執行兩格程式碼：先設定繁體中文字型，再顯示資料表並畫圖。

```python
%pip install -q mplfonts

import glob
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import mplfonts
from matplotlib import font_manager
from mplfonts import use_font

# 沿用本專案其他 Colab 範例的中文字型設定；此字型可顯示繁體中文。
for font_file in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(font_file)
sns.set_theme(style="whitegrid")
use_font("Noto Sans CJK SC")
plt.rcParams["axes.unicode_minus"] = False
print("✅ 繁體中文圖表環境準備完成")
```

```python
df = pd.DataFrame({"季度": ["第 1 季", "第 2 季", "第 3 季", "第 4 季"],
                   "營收（萬元）": [200, 225, 240, 280]})
display(df)
fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(df["季度"], df["營收（萬元）"], color=["#93c5fd", "#60a5fa", "#3b82f6", "#1d4ed8"])
ax.bar_label(bars, padding=3)
ax.set(title="四季營收逐季增加", ylabel="營收（萬元）", ylim=(0, 310))
plt.tight_layout()
plt.show()
```

## 看圖表說商業分析

**觀察：**第 4 季 280 萬元最高，比第 1 季 200 萬元多 80 萬元、增加 40%。

**商業解讀：**可提早規畫第 4 季的庫存與服務人力。

**限制：**季度總額成長不等於每月都成長；若想看季內波動，需再看月資料。
