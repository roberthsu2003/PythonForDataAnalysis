**何時選這張圖：**三個以上數值欄位，快速比較兩兩相關程度。

本例使用少量**課堂示意資料**，直接寫在程式中。依序執行兩格程式碼：先設定繁體中文字型，再顯示資料表並畫圖。

```python
%pip install -q mplfonts

import glob
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import mplfonts
from matplotlib import font_manager
from mplfonts import use_font

# 沿用本專案其他 Colab 範例的中文字型設定；此字型可顯示繁體中文。
for font_file in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(font_file)
sns.set_theme(style="whitegrid")
use_font("Noto Sans CJK SC")
plt.rcParams["axes.unicode_minus"] = False
print("✅ 繁體中文圖表環境準備完成")
```

```python
df = pd.DataFrame({"廣告費": [10, 20, 30, 40, 50, 60, 70, 80],
                   "訪客數": [120, 170, 180, 200, 310, 290, 350, 340],
                   "訂單數": [12, 20, 17, 26, 24, 35, 30, 39],
                   "營收": [24, 43, 40, 58, 55, 66, 74, 76],
                   "退貨數": [3, 4, 2, 5, 2, 6, 4, 3]})
display(df)
corr = df.corr(numeric_only=True)
fig, ax = plt.subplots(figsize=(7, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="Blues", vmin=-1, vmax=1,
            square=True, linewidths=0.5, ax=ax)
ax.set_title("營運指標的相關係數矩陣")
plt.tight_layout()
plt.show()
```

## 看圖表說商業分析

**觀察：**廣告費、訪客數、訂單數與營收多呈正相關；退貨數和訪客數的相關係數約 0.09，沒有明顯線性關聯。對角線永遠是 1。

**商業解讀：**可挑訪客數與訂單數，進一步檢查轉換率是否穩定。

**限制：**只有 8 個月份，且可能共同受季節影響；高相關仍不能證明廣告費造成營收增加。
